# Session 4: Measures of Dispersion

Covers Range, Variance, Standard Deviation, and Coefficient of Variation (CV) using everyday examples — step counts, Zomato orders, mobile data usage, Instagram likes, and UPI transactions.

## 1. Range of Daily Step Counts

**Range = Maximum value − Minimum value**

```
daily_steps = [6500, 7200, 8000, 6000, 9000, 7500, 8200]

step_range = max(daily_steps) - min(daily_steps)
print(f"Steps: {daily_steps}")
print(f"Max = {max(daily_steps)}, Min = {min(daily_steps)}")
print(f"Range = {max(daily_steps)} - {min(daily_steps)} = {step_range} steps")
```

**Interpretation:** A range of 3000 steps shows a noticeable swing in daily activity — the user isn't walking a consistent amount every day. Some days (like the 9000-step day) look much more active than others (like the 6000-step day), suggesting activity level depends on that day's routine (e.g. workday vs. rest day) rather than being a steady habit.

## 2. Variance and Standard Deviation of Zomato Order Amounts

Data: `[250, 400, 350, 300, 500]`

**Steps:** find the mean → find each deviation from the mean → square the deviations → average them (variance) → take the square root (standard deviation).

```
import math

order_amounts = [250, 400, 350, 300, 500]

def calculate_variance_std(values, population=True):
    n = len(values)
    mean_val = sum(values) / n
    squared_diffs = [(x - mean_val) ** 2 for x in values]
    divisor = n if population else (n - 1)
    variance = sum(squared_diffs) / divisor
    std_dev = math.sqrt(variance)
    return mean_val, variance, std_dev

mean_orders, variance_orders, std_orders = calculate_variance_std(order_amounts, population=True)

print(f"Order amounts: {order_amounts}")
print(f"Mean = {mean_orders}")
print(f"Variance (population) = {variance_orders}")
print(f"Standard Deviation (population) = {std_orders:.2f}")
```

## 3. Coefficient of Variation (CV) of Mobile Data Usage

Data (GB/month): `[3.2, 2.8, 3.5, 3.0, 2.9, 3.1]`

**CV = (Standard Deviation / Mean) × 100%**

```
data_usage = [3.2, 2.8, 3.5, 3.0, 2.9, 3.1]

mean_usage, variance_usage, std_usage = calculate_variance_std(data_usage, population=True)
cv_usage = (std_usage / mean_usage) * 100

print(f"Data usage (GB): {data_usage}")
print(f"Mean = {mean_usage:.4f} GB")
print(f"Standard Deviation = {std_usage:.4f} GB")
print(f"CV = ({std_usage:.4f} / {mean_usage:.4f}) x 100 = {cv_usage:.2f}%")
```

**Interpretation:** A CV of roughly 7% is quite low — as a rough rule of thumb, a CV under ~15-20% is considered 'consistent'. This means the user's monthly data usage is **fairly consistent** month to month, with only small fluctuations around the average of ~3.08 GB.

## 4. Standard Deviation of Instagram Likes

Data: `[120, 150, 130, 110, 170, 140, 135, 160, 125, 155]`

```
likes = [120, 150, 130, 110, 170, 140, 135, 160, 125, 155]

mean_likes, variance_likes, std_likes = calculate_variance_std(likes, population=True)
cv_likes = (std_likes / mean_likes) * 100

print(f"Likes: {likes}")
print(f"Mean = {mean_likes}")
print(f"Variance = {variance_likes}")
print(f"Standard Deviation = {std_likes:.2f}")
print(f"CV = {cv_likes:.2f}%")
```

**Comment on consistency:** The standard deviation (~18.1 likes) is about **13% of the mean** (139.5 likes). This is a moderate spread — engagement isn't wildly erratic, but there's a meaningful gap between the best-performing post (170 likes) and the weakest (110 likes), so engagement varies noticeably from post to post rather than being tightly consistent.

## 5. Variance & Standard Deviation of a Generated UPI Transaction Dataset

Below is a sample week of daily UPI transaction amounts (in ₹), generated as a representative example of realistic day-to-day spending (mix of small and large transactions).

```
upi_transactions = [220, 350, 180, 500, 275, 190, 410, 300]  # 8 days of UPI spends (₹)

mean_upi, variance_upi, std_upi = calculate_variance_std(upi_transactions, population=True)
cv_upi = (std_upi / mean_upi) * 100

print(f"UPI transactions (₹): {upi_transactions}")
print(f"Mean = ₹{mean_upi:.2f}")
print(f"Variance = {variance_upi:.2f}")
print(f"Standard Deviation = ₹{std_upi:.2f}")
print(f"CV = {cv_upi:.2f}%")
```

**Interpretation:** With a CV of roughly 35%, this user's UPI spending is **fairly variable** rather than stable — the standard deviation (~₹105) is a sizable fraction of the average transaction (~₹303). This pattern usually means spending includes a mix of small routine payments (e.g. ₹180-220) and occasional larger ones (e.g. ₹500), rather than a fixed daily spend. For someone budgeting, this level of variability suggests it's worth tracking which days drive the bigger transactions rather than assuming a flat daily average.
