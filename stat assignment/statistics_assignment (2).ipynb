```python
# Task 1: Daily step counts for last 7 days - calculate the range
import numpy as np

steps = [6500, 7200, 8000, 6000, 9000, 7500, 8200]
steps_range = max(steps) - min(steps)
print("Range of steps:", steps_range)
```

```python
# Task 2: 5 Zomato order amounts - compute variance and standard deviation
import numpy as np

orders = [250, 400, 350, 300, 500]
variance = np.var(orders)
std_dev = np.std(orders)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

```python
# Task 3: Monthly mobile data usage (6 months) - calculate coefficient of variation (CV)
import numpy as np

data_usage = [3.2, 2.8, 3.5, 3.0, 2.9, 3.1]
mean_usage = np.mean(data_usage)
std_usage = np.std(data_usage)
cv = (std_usage / mean_usage) * 100
print("Standard Deviation:", std_usage)
print("Coefficient of Variation:", cv, "%")
```

```python
# Task 4: Likes on 10 Instagram posts - compute standard deviation and compare with mean
import numpy as np

likes = [120, 150, 130, 110, 170, 140, 135, 160, 125, 155]
mean_likes = np.mean(likes)
std_likes = np.std(likes)
print("Mean:", mean_likes)
print("Standard Deviation:", std_likes)
```
