```python
# Task 1: Contingency table - favorite food delivery app vs preferred payment method (sample of 20+ responses)
import pandas as pd

responses = {
    'app':     ['Zomato','Zomato','Zomato','Zomato','Zomato','Zomato','Zomato',
                'Swiggy','Swiggy','Swiggy','Swiggy','Swiggy','Swiggy','Swiggy',
                'Uber Eats','Uber Eats','Uber Eats','Uber Eats','Uber Eats','Uber Eats','Uber Eats'],
    'payment': ['Paytm','PhonePe','Credit Card','Paytm','PhonePe','Paytm','Credit Card',
                'PhonePe','Paytm','Credit Card','PhonePe','Paytm','PhonePe','Credit Card',
                'Credit Card','Paytm','PhonePe','Credit Card','Paytm','PhonePe','Credit Card']
}

df = pd.DataFrame(responses)
contingency_table = pd.crosstab(df['app'], df['payment'])
print(contingency_table)
```

```python
# Task 2: Expected counts for music streaming preference (Spotify/Gaana/YouTube Music) by gender, assuming independence
import pandas as pd

observed = pd.DataFrame({
    'Spotify': [12, 18],
    'Gaana': [8, 7],
    'YouTube Music': [10, 15]
}, index=['Male', 'Female'])

grand_total = observed.values.sum()
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)

expected = pd.DataFrame(index=observed.index, columns=observed.columns, dtype=float)
for row in observed.index:
    for col in observed.columns:
        expected.loc[row, col] = (row_totals[row] * col_totals[col]) / grand_total

print("Observed counts:")
print(observed)
print()
print("Expected counts:")
print(expected)
```

```python
# Task 3: Chi-square test of independence - app choice vs payment method (using contingency table from Task 1)
import pandas as pd
from scipy.stats import chi2_contingency

responses = {
    'app':     ['Zomato','Zomato','Zomato','Zomato','Zomato','Zomato','Zomato',
                'Swiggy','Swiggy','Swiggy','Swiggy','Swiggy','Swiggy','Swiggy',
                'Uber Eats','Uber Eats','Uber Eats','Uber Eats','Uber Eats','Uber Eats','Uber Eats'],
    'payment': ['Paytm','PhonePe','Credit Card','Paytm','PhonePe','Paytm','Credit Card',
                'PhonePe','Paytm','Credit Card','PhonePe','Paytm','PhonePe','Credit Card',
                'Credit Card','Paytm','PhonePe','Credit Card','Paytm','PhonePe','Credit Card']
}

df = pd.DataFrame(responses)
contingency_table = pd.crosstab(df['app'], df['payment'])

chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print("Contingency Table:")
print(contingency_table)
print()
print("Chi-square statistic:", chi2_stat)
print("Degrees of freedom:", dof)
print("P-value:", p_value)
print()
print("Expected counts:")
print(expected)

alpha = 0.05
if p_value < alpha:
    print("\nConclusion: Reject the null hypothesis - there IS a significant association between app choice and payment method.")
else:
    print("\nConclusion: Fail to reject the null hypothesis - no significant association between app choice and payment method.")
```

```python
# Task 4: Chi-square goodness of fit test - Flipkart claims 70% will prefer the new product (30 users surveyed)
from scipy.stats import chisquare

observed = [18, 12]  # prefer, do not prefer
n = sum(observed)

claimed_proportion_prefer = 0.70
expected = [n * claimed_proportion_prefer, n * (1 - claimed_proportion_prefer)]

chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

print("Observed counts:", observed)
print("Expected counts:", expected)
print("Chi-square statistic:", chi2_stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Conclusion: Reject the null hypothesis - the observed preference significantly differs from the claimed 70%.")
else:
    print("Conclusion: Fail to reject the null hypothesis - the data is consistent with the claim that 70% prefer the new product.")
```
