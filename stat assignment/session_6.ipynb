## 1. Sample Space: Rolling a Die and Flipping a Coin

When you roll a standard six-sided die **and** flip a coin at the same time, each outcome is an ordered pair `(die_face, coin_face)`. Since the die has 6 outcomes and the coin has 2, the sample space has `6 × 2 = 12` equally likely outcomes.

```
# Generate the sample space programmatically
die_faces = range(1, 7)
coin_faces = ['H', 'T']

sample_space = [(d, c) for d in die_faces for c in coin_faces]
print(f"Total outcomes: {len(sample_space)}\n")
for outcome in sample_space:
    print(outcome)
```

**Sample space (12 outcomes):**

(1,H), (1,T), (2,H), (2,T), (3,H), (3,T), (4,H), (4,T), (5,H), (5,T), (6,H), (6,T)

## 2. Classical Probability: Drawing an Even-Numbered Card

Deck: 10 cards numbered 1 to 10.

**Step 1 — Total outcomes:** n(S) = 10

**Step 2 — Favorable outcomes (even numbers):** {2, 4, 6, 8, 10} → n(E) = 5

**Step 3 — Classical probability formula:**

$$P(E) = \frac{n(E)}{n(S)} = \frac{5}{10} = 0.5$$

```
deck = list(range(1, 11))
even_cards = [n for n in deck if n % 2 == 0]

n_total = len(deck)
n_even = len(even_cards)
p_even = n_even / n_total

print(f"Even-numbered cards: {even_cards}")
print(f"n(S) = {n_total}, n(E) = {n_even}")
print(f"P(Even) = {n_even}/{n_total} = {p_even}")
```

## 3. Probability a Random Song is NOT by Arijit Singh

Playlist: 8 songs total, 3 by Arijit Singh → 5 songs are NOT by Arijit Singh.

**Step 1:** P(Arijit) = 3/8

**Step 2:** P(Not Arijit) = 1 − P(Arijit) = 1 − 3/8 = 5/8

**Step 3:** 5/8 = 0.625

```
total_songs = 8
arijit_songs = 3
not_arijit_songs = total_songs - arijit_songs

p_arijit = arijit_songs / total_songs
p_not_arijit = 1 - p_arijit

print(f"P(Arijit) = {arijit_songs}/{total_songs} = {p_arijit}")
print(f"P(Not Arijit) = 1 - {p_arijit} = {p_not_arijit}")
```

## 4. Addition Rule: Zomato User Ordering Pizza or Burger

Given:
- P(Pizza) = 0.3
- P(Burger) = 0.4
- P(Pizza and Burger) = 0.1

**Addition Rule:**

$$P(Pizza \text{ or } Burger) = P(Pizza) + P(Burger) - P(Pizza \text{ and } Burger)$$

$$= 0.3 + 0.4 - 0.1 = 0.6$$

```
p_pizza = 0.3
p_burger = 0.4
p_both = 0.1

p_pizza_or_burger = p_pizza + p_burger - p_both
print(f"P(Pizza or Burger) = {p_pizza} + {p_burger} - {p_both} = {p_pizza_or_burger}")
```

## 5. Conditional Probability: Flipkart Phone Add-to-Cart

Given:
- P(Add phone to cart) = 0.2 (overall, across all browsing users)
- P(Add phone to cart | Interested in mobiles) = 0.5

**Explanation (in plain words):**

Without any extra information, only 20% of browsing users end up adding a phone to their cart. But once we *condition* on the fact that a user is specifically browsing/interested in the mobiles category, that probability jumps to 50%. In other words, knowing the user's category interest gives us extra information that changes how likely the event (adding a phone to cart) is. This shows that 'adding a phone to cart' and 'being interested in mobiles' are **not independent** — interest in mobiles makes the phone purchase action much more likely, which makes intuitive sense: someone already looking at mobile-related content is a 'warmer' lead than a random browser.

```
p_add_phone = 0.2
p_add_phone_given_mobile_interest = 0.5

print(f"P(Add Phone to Cart) = {p_add_phone}")
print(f"P(Add Phone to Cart | Interested in Mobiles) = {p_add_phone_given_mobile_interest}")
print("Since these two values differ, the events are dependent -")
print("knowing the user's mobile interest changes the probability of the cart action.")
```
