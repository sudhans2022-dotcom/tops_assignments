# Session 3: Measures of Central Tendency

Covers Mean, Median, and Mode with everyday examples — step counts, delivery times, YouTube genres, and UPI transactions.

## 1. Mean of Daily Step Counts

Sample data: 7 days of step counts from a fitness app.

**Mean formula:**

$$\text{Mean} = \frac{\sum \text{steps}}{n}$$

```
daily_steps = [8200, 6500, 10300, 4800, 9100, 7600, 11200]

def calculate_mean(values):
    return sum(values) / len(values)

mean_steps = calculate_mean(daily_steps)
print(f"Daily steps: {daily_steps}")
print(f"Mean steps = sum({daily_steps}) / {len(daily_steps)}")
print(f"Mean steps = {sum(daily_steps)} / {len(daily_steps)} = {mean_steps:.2f}")
```

## 2. Median Delivery Time (Zomato/Swiggy)

Data: `[32, 28, 29, 45, 30, 31, 60, 30, 29]` (9 values)

**Steps:**
1. Sort the values.
2. Since n = 9 is odd, the median is the middle value at position `(n+1)/2`.

```
delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

def find_median(values):
    sorted_vals = sorted(values)
    n = len(sorted_vals)
    mid = n // 2
    if n % 2 == 1:
        # Odd count -> middle element
        return sorted_vals[mid]
    else:
        # Even count -> average the two middle elements
        return (sorted_vals[mid - 1] + sorted_vals[mid]) / 2

sorted_times = sorted(delivery_times)
median_time = find_median(delivery_times)

print(f"Original times: {delivery_times}")
print(f"Sorted times:   {sorted_times}")
print(f"n = {len(delivery_times)} (odd) -> middle position = {len(delivery_times)//2 + 1}")
print(f"Median delivery time = {median_time} minutes")
```

## 3. Mode of YouTube Video Genres

Data: 12 genres watched this week.

**Mode** = the value(s) that appear most frequently.

```
from collections import Counter

genres = ['Music', 'Vlog', 'Music', 'Tech', 'Music', 'Vlog', 'Tech', 'Music', 'Comedy', 'Vlog', 'Music', 'Comedy']

def find_mode(values):
    counts = Counter(values)
    max_count = max(counts.values())
    modes = [val for val, cnt in counts.items() if cnt == max_count]
    return modes, max_count, counts

modes, max_count, all_counts = find_mode(genres)

print(f"Genre counts: {dict(all_counts)}")
print(f"Mode (most watched genre): {modes[0]} (watched {max_count} times)")
```

## 4. Mean vs Median for UPI Transactions

Data: `[100, 120, 105, 110, 5000, 115, 108]`

One value (₹5000) is a clear outlier compared to the rest (all around ₹100–120).

```
upi_amounts = [100, 120, 105, 110, 5000, 115, 108]

mean_amount = calculate_mean(upi_amounts)
median_amount = find_median(upi_amounts)

print(f"Transactions: {upi_amounts}")
print(f"Sorted:       {sorted(upi_amounts)}")
print(f"Mean amount   = ₹{mean_amount:.2f}")
print(f"Median amount = ₹{median_amount}")
```

**Explanation:**

The **median (₹110)** is the better measure of the 'typical' transaction here. The ₹5000 entry is an extreme outlier that pulls the **mean up to ~₹794**, which is far higher than every single transaction except that one — so the mean doesn't represent a realistic 'typical' spend. The median, being based on rank/position rather than magnitude, isn't affected by that outlier and stays close to where most of the actual values cluster.
