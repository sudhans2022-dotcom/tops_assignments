```python
# Task 1: Simulate a Bernoulli trial - Zomato login success (p=0.85), run 20 times
import numpy as np

np.random.seed(42)
p = 0.85
n_trials = 20

results = np.random.binomial(1, p, n_trials)
print("Login results (1=success, 0=failure):", results)
print("Number of successes:", np.sum(results))
```

```python
# Task 2: Binomial distribution - probability of exactly 3 successful UPI payments out of 5 (p=0.9)
from scipy.stats import binom

def prob_exact_successes(n, k, p):
    return binom.pmf(k, n, p)

probability = prob_exact_successes(n=5, k=3, p=0.9)
print("P(exactly 3 successes out of 5):", probability)
```

```python
# Task 3: Poisson distribution - probability of exactly 6 playlist creations in a minute (avg = 4/min)
from scipy.stats import poisson

lam = 4
k = 6

probability = poisson.pmf(k, lam)
print("P(exactly 6 playlists created):", probability)
```

```python
# Task 4: Simulate and plot number of Flipkart reviews in 12 random hours using Poisson distribution (avg = 10/hour)
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
lam = 10
n_hours = 12

reviews_per_hour = np.random.poisson(lam, n_hours)
print("Reviews per hour:", reviews_per_hour)

plt.figure(figsize=(8, 5))
plt.bar(range(1, n_hours + 1), reviews_per_hour, color='skyblue', edgecolor='black')
plt.title("Simulated Product Reviews per Hour (Poisson, λ=10)")
plt.xlabel("Hour")
plt.ylabel("Number of Reviews")
plt.xticks(range(1, n_hours + 1))
plt.show()
```
