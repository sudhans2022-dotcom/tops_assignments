```python
# Task 1: Pearson correlation between danceability and popularity for 12 songs
import pandas as pd

songs_data = {
    'song': [f'Song_{i}' for i in range(1, 13)],
    'danceability': [0.72, 0.65, 0.80, 0.58, 0.90, 0.45, 0.68, 0.77, 0.55, 0.83, 0.60, 0.95],
    'energy': [0.65, 0.70, 0.85, 0.50, 0.88, 0.40, 0.60, 0.75, 0.52, 0.80, 0.58, 0.92],
    'popularity': [70, 60, 85, 45, 92, 35, 65, 78, 50, 88, 55, 95]
}

df = pd.DataFrame(songs_data)
pearson_corr = df['danceability'].corr(df['popularity'], method='pearson')

print(df)
print()
print("Pearson correlation (danceability vs popularity):", pearson_corr)
```

```python
# Task 2: Spearman correlation between IPL players' total runs and Instagram followers (in thousands)
from scipy.stats import spearmanr
import pandas as pd

ipl_data = {
    'player': [f'Player_{i}' for i in range(1, 11)],
    'total_runs': [5500, 6200, 3200, 4800, 7000, 2100, 4300, 5800, 3900, 6700],
    'instagram_followers_k': [15000, 22000, 3000, 9000, 35000, 1200, 6000, 18000, 5000, 28000]
}

df = pd.DataFrame(ipl_data)
spearman_corr, p_value = spearmanr(df['total_runs'], df['instagram_followers_k'])

print(df)
print()
print("Spearman correlation:", spearman_corr)
print("P-value:", p_value)
print()
print("Interpretation: A strong positive Spearman correlation indicates that players with higher total runs "
      "generally tend to have more Instagram followers, i.e. on-field performance ranking is associated "
      "with social media popularity ranking.")
```

```python
# Task 3: Correlation matrix for Flipkart product features - price, user rating, number of reviews
import pandas as pd

flipkart_data = {
    'price': [1200, 2500, 800, 3200, 1500, 4500, 950, 2100, 3800, 1700],
    'user_rating': [4.2, 4.5, 3.8, 4.7, 4.0, 4.8, 3.9, 4.3, 4.6, 4.1],
    'num_reviews': [340, 620, 150, 890, 410, 1200, 180, 500, 950, 380]
}

df = pd.DataFrame(flipkart_data)
correlation_matrix = df.corr()

print(correlation_matrix)
```

```python
# Task 4: Visualize the Flipkart correlation matrix as a heatmap using seaborn
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

flipkart_data = {
    'price': [1200, 2500, 800, 3200, 1500, 4500, 950, 2100, 3800, 1700],
    'user_rating': [4.2, 4.5, 3.8, 4.7, 4.0, 4.8, 3.9, 4.3, 4.6, 4.1],
    'num_reviews': [340, 620, 150, 890, 410, 1200, 180, 500, 950, 380]
}

df = pd.DataFrame(flipkart_data)
correlation_matrix = df.corr()

plt.figure(figsize=(6, 5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', cbar=True, fmt=".2f")
plt.title("Correlation Heatmap: Price, Rating, Reviews")
plt.show()
```

```python
# Task 5: Two potential pitfalls when interpreting correlation coefficients (Zomato/YouTube examples)

pitfalls = '''
Pitfall 1 - Correlation does not imply causation:
On Zomato, restaurants with more photos uploaded might show a strong positive correlation with higher
order counts, but that doesn't mean uploading more photos *causes* more orders - it could be that
already-popular restaurants naturally attract more customer-uploaded photos, reversing the assumed
direction of cause and effect.

Pitfall 2 - Confounding variables (hidden third factor):
On YouTube, video length might show a positive correlation with total watch time, but a hidden
confounder like "channel popularity" could be driving both - popular channels tend to post longer
videos AND get more watch time regardless of length, making the length-watch time correlation
misleading if length itself isn't the real driver.
'''
print(pitfalls)
```
