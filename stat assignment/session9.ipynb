```python
# Task 1: Generate 1000 random numbers from a normal distribution (mean=500, std=100) and plot
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.normal(loc=500, scale=100, size=1000)

plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title("Distribution of 1000 Random Numbers (Normal, μ=500, σ=100)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# Task 2: Calculate the Z-score for a day with 9500 steps (mean=8000, std=1200)
X = 9500
mu = 8000
sigma = 1200

z_score = (X - mu) / sigma
print("Z-score:", z_score)
```

```python
# Task 3: Plot standard normal curve and shade area between Z = -1 and Z = 1
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

z = np.linspace(-4, 4, 1000)
pdf = norm.pdf(z, 0, 1)

plt.figure(figsize=(8, 5))
plt.plot(z, pdf, color='navy')

z_fill = np.linspace(-1, 1, 500)
plt.fill_between(z_fill, norm.pdf(z_fill, 0, 1), color='skyblue', alpha=0.6)

plt.title("Standard Normal Distribution (Shaded: -1 ≤ Z ≤ 1)")
plt.xlabel("Z")
plt.ylabel("Probability Density")
plt.show()

prob_within_1_std = norm.cdf(1) - norm.cdf(-1)
print("Probability within 1 standard deviation:", prob_within_1_std)
```

```python
# Task 4: Probability a Zomato delivery takes less than 25 minutes (mean=30, std=5)
from scipy.stats import norm

mu = 30
sigma = 5
X = 25

probability = norm.cdf(X, mu, sigma)
print("P(Delivery time < 25 minutes):", probability)
```
