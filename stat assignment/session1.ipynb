# Session 1: Population vs Sample — Zomato Order Amounts

This notebook simulates the Excel/Google Sheets exercise in Python: entering 20 simulated Zomato order amounts, computing the mean/median/mode (equivalent to the `AVERAGE`, `MEDIAN`, and `MODE` spreadsheet functions), comparing a 5-value sample to the full population, and reflecting on the difference between the two.

## 1. Simulated 'Order Amount' Column (20 values, ₹100–₹1000)

In Excel/Sheets this would be a column named **Order Amount** with 20 rows filled in manually or with `=RANDBETWEEN(100,1000)`. Here we hard-code 20 representative values (with a few repeated amounts, since real Zomato orders often cluster around common price points like ₹250 or ₹430).

```
order_amount = [250, 480, 320, 610, 250, 999, 175, 430, 720, 250,
                560, 890, 340, 610, 480, 275, 999, 150, 430, 700]

print("Order Amount column (20 values):")
for i, val in enumerate(order_amount, start=1):
    print(f"Row {i:2d}: ₹{val}")
```

## 2. Mean, Median, and Mode (Population)

Spreadsheet equivalents:
- `=AVERAGE(A2:A21)` → Mean
- `=MEDIAN(A2:A21)` → Median
- `=MODE(A2:A21)` → Mode

```
import statistics

population_mean = statistics.mean(order_amount)          # like =AVERAGE(...)
population_median = statistics.median(order_amount)      # like =MEDIAN(...)
population_mode = statistics.mode(order_amount)          # like =MODE(...)

print(f"Mean (AVERAGE)  = ₹{population_mean:.2f}")
print(f"Median (MEDIAN) = ₹{population_median}")
print(f"Mode (MODE)     = ₹{population_mode}")
```

## 3. Sample Mean (First 5 Values) vs Population Mean

Taking the first 5 rows as a *sample* of the full 20-row *population*.

```
sample = order_amount[:5]
sample_mean = statistics.mean(sample)  # =AVERAGE(A2:A6)

print(f"Sample (first 5 values): {sample}")
print(f"Sample mean      = ₹{sample_mean:.2f}")
print(f"Population mean  = ₹{population_mean:.2f}")
print(f"Difference       = ₹{abs(population_mean - sample_mean):.2f}")
```

## 4. Population vs Sample — Explanation

In this exercise, the **population** is the complete set of all 20 Zomato order amounts recorded for the day — every single order that happened. The **sample** is a smaller subset drawn from that population, in this case just the first 5 orders. When we compared the two, the sample mean (₹382.00) came out noticeably lower than the population mean (₹495.90), simply because those first 5 orders happened to miss some of the larger values (like ₹999 and ₹890) that appear later in the day. This is the core risk of sampling: a small or non-random subset can give a misleading picture of the whole. In practice, a food delivery company would almost always prefer working with the **population** (or a large, randomly-selected sample) when it can, since that gives a far more accurate and reliable estimate of true average order value — a small, biased sample like this one is only useful as a quick, rough estimate when collecting the full population isn't feasible (e.g. analyzing orders in real time before the day is over).
