```python
# Task 1: One-sample t-test - is average Zomato delivery time significantly different from 30 minutes?
from scipy import stats

delivery_times = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]
population_mean = 30

t_stat, p_value = stats.ttest_1samp(delivery_times, population_mean)

print("Sample mean:", sum(delivery_times) / len(delivery_times))
print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: Reject the null hypothesis - delivery time is significantly different from 30 minutes.")
else:
    print("Result: Fail to reject the null hypothesis - no significant difference from 30 minutes.")
```

```python
# Task 2: Independent two-sample t-test - daily steps before vs after installing a fitness app
import numpy as np
from scipy import stats

np.random.seed(42)
steps_before = np.random.normal(loc=6000, scale=800, size=30)
steps_after = np.random.normal(loc=7200, scale=800, size=30)

t_stat, p_value = stats.ttest_ind(steps_after, steps_before)

print("Mean steps before:", np.mean(steps_before))
print("Mean steps after:", np.mean(steps_after))
print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: Reject the null hypothesis - average steps significantly increased after using the app.")
else:
    print("Result: Fail to reject the null hypothesis - no significant increase in steps.")
```

```python
# Task 3: Paired t-test - screen time (hours) before and after enabling Focus Mode, for 10 users
import numpy as np
from scipy import stats

np.random.seed(1)
screen_time_before = np.array([6.5, 7.2, 5.8, 8.0, 6.1, 7.5, 6.8, 7.0, 6.3, 7.8])
screen_time_after = screen_time_before - np.random.normal(loc=1.0, scale=0.4, size=10)

t_stat, p_value = stats.ttest_rel(screen_time_before, screen_time_after)

print("Mean screen time before:", np.mean(screen_time_before))
print("Mean screen time after:", np.mean(screen_time_after))
print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: Reject the null hypothesis - Focus Mode significantly reduced screen time.")
else:
    print("Result: Fail to reject the null hypothesis - no significant reduction in screen time.")
```

```python
# Task 4: Compare average ratings of Zomato vs Swiggy (15 ratings each) using the appropriate t-test
import numpy as np
from scipy import stats

np.random.seed(5)
zomato_ratings = np.random.normal(loc=4.2, scale=0.4, size=15)
swiggy_ratings = np.random.normal(loc=4.0, scale=0.4, size=15)

# Independent two-sample t-test is used because the two samples (Zomato users and Swiggy users)
# are two separate, unrelated groups of ratings, not paired/matched observations.
t_stat, p_value = stats.ttest_ind(zomato_ratings, swiggy_ratings)

print("Mean Zomato rating:", np.mean(zomato_ratings))
print("Mean Swiggy rating:", np.mean(swiggy_ratings))
print("T-statistic:", t_stat)
print("P-value:", p_value)

reason = ("An independent (two-sample) t-test is appropriate here because the Zomato ratings and "
          "Swiggy ratings come from two different, independent groups of users rather than the same "
          "users rating both apps.")
print("Reason for test choice:", reason)
```

```python
# Task 5: One-sample t-test - average daily UPI transactions per user on PhonePe
import numpy as np
from scipy import stats

np.random.seed(10)
daily_upi_transactions = np.random.normal(loc=3.5, scale=1.2, size=25)  # number of transactions per day per user
hypothesized_mean = 3.0

t_stat, p_value = stats.ttest_1samp(daily_upi_transactions, hypothesized_mean)

print("Sample data:", np.round(daily_upi_transactions, 2))
print("Sample mean:", np.mean(daily_upi_transactions))
print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: Reject the null hypothesis - average daily UPI transactions differ significantly from 3.0.")
else:
    print("Result: Fail to reject the null hypothesis - no significant difference from 3.0.")
```
