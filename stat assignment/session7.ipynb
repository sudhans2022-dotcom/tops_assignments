```python
# Task 1: P(Phone | Accessories) - e-commerce app, 500 users, 120 bought accessories, 80 bought phone, 50 bought both
total_users = 500
bought_accessories = 120
bought_phone = 80
bought_both = 50

p_accessories = bought_accessories / total_users
p_both = bought_both / total_users

p_phone_given_accessories = p_both / p_accessories

print("P(Accessories):", p_accessories)
print("P(Phone and Accessories):", p_both)
print("P(Phone | Accessories):", p_phone_given_accessories)
```

```python
# Task 2: P(Smartphone | Headphones) - Flipkart users, 30% buy headphones, 40% of those buy a smartphone
p_headphones = 0.30
p_smartphone_given_headphones = 0.40

# Using P(A|B) = P(A and B) / P(B) -> here 40% is already given directly as P(Smartphone|Headphones)
p_smartphone_and_headphones = p_smartphone_given_headphones * p_headphones

print("P(Headphones):", p_headphones)
print("P(Smartphone and Headphones):", p_smartphone_and_headphones)
print("P(Smartphone | Headphones):", p_smartphone_given_headphones)
```

```python
# Task 3: Bayes' Theorem function to compute churn probability given a 'we miss you' email
def bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email):
    # P(Churn | Email) = [P(Email | Churn) * P(Churn)] / P(Email)
    return (prob_email_given_churn * prior_churn) / prob_email

result = bayes_churn_probability(prior_churn=0.15, prob_email_given_churn=0.6, prob_email=0.25)
print("P(Churn | Email):", result)
```

```python
# Task 4: Bayes' Theorem - probability a transaction is actually fraudulent when flagged by the alert
p_fraud = 0.02
p_not_fraud = 1 - p_fraud

p_alert_given_fraud = 0.90
p_alert_given_not_fraud = 0.05

# P(A) = P(A|F)*P(F) + P(A|not F)*P(not F)
p_alert = (p_alert_given_fraud * p_fraud) + (p_alert_given_not_fraud * p_not_fraud)

p_fraud_given_alert = (p_alert_given_fraud * p_fraud) / p_alert

print("P(Alert):", p_alert)
print("P(Fraud | Alert):", p_fraud_given_alert)
```
