```python
# Task 1: Create purchase_data.csv with customer_id, age_group, purchase_value (5 customers per age group, 15 total)
import pandas as pd

data = {
    'customer_id': list(range(1, 16)),
    'age_group': ['18-25']*5 + ['26-35']*5 + ['36-50']*5,
    'purchase_value': [
        450, 500, 480, 520, 470,      # 18-25
        700, 650, 720, 680, 710,      # 26-35
        900, 950, 880, 920, 940       # 36-50
    ]
}

df = pd.DataFrame(data)
df.to_csv('purchase_data.csv', index=False)
print(df)
```

```python
# Task 2: One-way ANOVA - compare average purchase_value across the three age groups (reads purchase_data.csv)
import pandas as pd
from scipy.stats import f_oneway

df = pd.read_csv('purchase_data.csv')

group_18_25 = df[df['age_group'] == '18-25']['purchase_value']
group_26_35 = df[df['age_group'] == '26-35']['purchase_value']
group_36_50 = df[df['age_group'] == '36-50']['purchase_value']

f_stat, p_value = f_oneway(group_18_25, group_26_35, group_36_50)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
# Task 3: Interpret the ANOVA results

import pandas as pd
from scipy.stats import f_oneway

df = pd.read_csv('purchase_data.csv')
groups = [g['purchase_value'].values for _, g in df.groupby('age_group')]
f_stat, p_value = f_oneway(*groups)

alpha = 0.05
print("F-statistic:", f_stat)
print("P-value:", p_value)

interpretation = (
    "Since the p-value is far smaller than 0.05, we reject the null hypothesis. "
    "This means there IS a statistically significant difference in average purchase_value "
    "between the three age groups (18-25, 26-35, 36-50)."
)
print(interpretation)
```

```python
# Task 4: Calculate between-group variance and within-group variance for purchase_value
import pandas as pd
import numpy as np

df = pd.read_csv('purchase_data.csv')
grand_mean = df['purchase_value'].mean()

groups = df.groupby('age_group')['purchase_value']

# Between-group variance (weighted variance of group means around the grand mean)
between_group_var = sum(
    len(group) * (group.mean() - grand_mean) ** 2 for _, group in groups
) / (len(groups) - 1)

# Within-group variance (average variance of values around their own group mean)
within_group_var = sum(
    ((group - group.mean()) ** 2).sum() for _, group in groups
) / (len(df) - len(groups))

print("Between-group variance:", between_group_var)
print("Within-group variance:", within_group_var)

explanation = ("Between-group variance measures how much the average purchase_value differs "
               "across the three age groups. Within-group variance measures how much purchase_value "
               "varies among customers inside the same age group.")
print(explanation)
```

```python
# Task 5: One-way ANOVA on Zomato monthly order amounts for Students, Working Professionals, Retirees (dummy data)
import numpy as np
from scipy.stats import f_oneway

np.random.seed(42)

students = np.random.normal(loc=1200, scale=200, size=20)
working_professionals = np.random.normal(loc=2500, scale=300, size=20)
retirees = np.random.normal(loc=1800, scale=250, size=20)

f_stat, p_value = f_oneway(students, working_professionals, retirees)

print("Students mean:", np.mean(students))
print("Working Professionals mean:", np.mean(working_professionals))
print("Retirees mean:", np.mean(retirees))
print()
print("F-statistic:", f_stat)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Conclusion: Significant difference in average monthly order amounts between the segments.")
else:
    print("Conclusion: No significant difference in average monthly order amounts between the segments.")
```
