```python
# Task 1: State null and alternative hypotheses - cashback offer effect on average daily orders (Zomato)

null_hypothesis = "H0: The average number of daily orders after the cashback offer is equal to (or less than or equal to) the average before the offer (mu_after <= mu_before)."
alternative_hypothesis = "H1: The average number of daily orders after the cashback offer is greater than the average before the offer (mu_after > mu_before)."

print("Null Hypothesis (H0):", null_hypothesis)
print("Alternative Hypothesis (H1):", alternative_hypothesis)
```

```python
# Task 2: Identify error type - wrongly concluding the recommendation algorithm works when it actually doesn't

error_type = "Type I Error"
explanation = ("This is a Type I Error because it means rejecting a true null hypothesis - "
               "concluding the algorithm increases listening time (from 40 to 45 minutes) "
               "when in reality it has no real effect.")

print("Error Type:", error_type)
print("Explanation:", explanation)
```

```python
# Task 3: A/B test decision - Flipkart discount banner, p-value=0.03, 95% confidence level (alpha=0.05)

p_value = 0.03
alpha = 0.05

if p_value < alpha:
    decision = "Reject the null hypothesis"
else:
    decision = "Fail to reject the null hypothesis"

interpretation = ("Since the p-value (0.03) is less than alpha (0.05), we reject the null hypothesis. "
                   "This means there is statistically significant evidence that the new discount banner "
                   "does increase purchases on the app, so Flipkart could consider rolling it out more broadly.")

print("Decision:", decision)
print("Interpretation:", interpretation)
```

```python
# Task 4: What does a 99% confidence level mean for a cricket app's push notification feature?

explanation = '''
A 99% confidence level means that if this test were repeated many times, 99% of the resulting
confidence intervals would contain the true effect of match alerts on user engagement - i.e., the app
is being very strict about ruling out random chance before concluding the notifications genuinely help.
This higher confidence level (compared to the more common 95%) requires stronger evidence before
rejecting the null hypothesis, so the corresponding significance level (alpha = 0.01) is smaller. As a
result, the app will only accept and roll out the push notification feature if the observed increase in
engagement is very unlikely to have happened by chance, reducing the risk of a Type I Error (falsely
believing the feature works) but making it a bit harder to detect a real but smaller effect (raising the
risk of a Type II Error).
'''
print(explanation)
```
