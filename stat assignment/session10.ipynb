```python
# Task 1: Random sample of 20 users from a list of 200 user IDs (Spotify feedback survey)
import random

random.seed(42)
user_ids = [f"user_{i}" for i in range(1, 201)]

sample = random.sample(user_ids, 20)
print("Sampled users:", sample)
```

```python
# Task 2: Stratified sample of 60 orders from 1000 Flipkart orders labeled by city
import random
import pandas as pd

random.seed(42)
cities = ['Ahmedabad', 'Surat', 'Vadodara']
weights = [0.5, 0.3, 0.2]  # proportion of total orders coming from each city

order_cities = random.choices(cities, weights=weights, k=1000)
df = pd.DataFrame({'order_id': range(1, 1001), 'city': order_cities})

sample_size = 60
stratified_sample = pd.DataFrame()

for city in cities:
    city_group = df[df['city'] == city]
    proportion = len(city_group) / len(df)
    n_city_sample = round(proportion * sample_size)
    stratified_sample = pd.concat([stratified_sample, city_group.sample(n_city_sample, random_state=42)])

print(stratified_sample['city'].value_counts())
print("Total sampled orders:", len(stratified_sample))
```

```python
# Task 3: Systematic sampling - select every 10th review from 500 Zomato restaurant reviews
reviews = [f"review_{i}" for i in range(1, 501)]

def systematic_sample(data, step):
    return data[::step]

sample = systematic_sample(reviews, 10)
print("Sampled reviews:", sample)
print("Number of reviews sampled:", len(sample))
```

```python
# Task 4: Effect of sample size on accuracy of estimated average Swiggy delivery time (10,000 records)
import numpy as np

np.random.seed(42)
# Simulated population of 10,000 delivery times (minutes), true mean around 30
population = np.random.normal(loc=30, scale=6, size=10000)
true_mean = np.mean(population)

for sample_size in [50, 200, 500]:
    sample = np.random.choice(population, sample_size, replace=False)
    sample_mean = np.mean(sample)
    std_error = np.std(sample) / np.sqrt(sample_size)
    print(f"Sample size: {sample_size}")
    print(f"  Sample mean: {sample_mean:.3f}")
    print(f"  Standard Error: {std_error:.3f}")
    print(f"  Difference from true mean ({true_mean:.3f}): {abs(sample_mean - true_mean):.3f}")
    print()
```

```python
# Task 5: Stratified sample of 40 IPL fans by favorite team, proportionally represented (own logic, no demo code reused)
import random

random.seed(7)

teams = ['CSK', 'MI', 'RCB', 'KKR', 'GT']
fan_counts = {'CSK': 350, 'MI': 300, 'RCB': 250, 'KKR': 60, 'GT': 40}
total_fans = sum(fan_counts.values())

# Build the full fan pool grouped by team
fan_pool = {}
fan_id = 1
for team, count in fan_counts.items():
    fan_pool[team] = []
    for _ in range(count):
        fan_pool[team].append(f"fan_{fan_id}")
        fan_id += 1

sample_size = 40
final_sample = {}

for team, fans in fan_pool.items():
    team_share = fan_counts[team] / total_fans
    n_from_team = round(team_share * sample_size)
    final_sample[team] = random.sample(fans, n_from_team)

for team, sampled_fans in final_sample.items():
    print(f"{team}: {len(sampled_fans)} fans sampled")

print("Total fans sampled:", sum(len(v) for v in final_sample.values()))
```
