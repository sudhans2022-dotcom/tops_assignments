```python
# Task 1: Simulate CLT - 1000 samples (size 30) from a skewed dataset (Instagram likes), plot histogram of sample means
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Skewed population: number of likes on Instagram posts (exponential-like skew)
population = np.random.exponential(scale=150, size=100000)

sample_means = []
for _ in range(1000):
    sample = np.random.choice(population, size=30, replace=True)
    sample_means.append(np.mean(sample))

plt.figure(figsize=(8, 5))
plt.hist(sample_means, bins=30, color='lightgreen', edgecolor='black')
plt.title("Distribution of Sample Means (n=30) from Skewed Instagram Likes Data")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```

```python
# Task 2: Explanation - why the sampling distribution of the mean becomes approximately normal (example: Zomato order amounts)

explanation = '''
Individual Zomato order amounts are usually skewed - most orders are small or medium sized, with a
few very large orders pulling the distribution's tail to the right. However, if we repeatedly take
random samples of, say, 30 orders and calculate the average order amount for each sample, the
Central Limit Theorem tells us that the distribution of these sample averages will look approximately
normal (bell-shaped), even though the original order-amount data is skewed. This happens because
averaging many random values tends to cancel out extreme highs and lows, so the sample means cluster
more tightly and symmetrically around the true population mean as the sample size grows.
'''
print(explanation)
```

```python
# Task 3: Real-world scenario where CLT helps analytics/business decisions (Swiggy)

scenario = '''
Swiggy wants to estimate the average delivery time across all orders in a city to decide whether it
needs to add more delivery partners during peak hours. Since delivery times vary a lot and the raw
distribution can be skewed (a few orders take unusually long due to traffic or distance), Swiggy can
take repeated random samples of delivery times (e.g., 50 orders per sample) throughout the day. By the
Central Limit Theorem, the distribution of these sample means will be approximately normal regardless
of the shape of the underlying delivery-time data. This lets Swiggy confidently apply standard
statistical methods (like confidence intervals) to estimate the true average delivery time and detect
when it's significantly worse than usual, without needing to analyze every single order in the city.
'''
print(scenario)
```

```python
# Task 4: Compare CLT simulation for small sample size (n=5) vs n=30
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
population = np.random.exponential(scale=150, size=100000)

sample_means_n5 = [np.mean(np.random.choice(population, size=5, replace=True)) for _ in range(1000)]
sample_means_n30 = [np.mean(np.random.choice(population, size=30, replace=True)) for _ in range(1000)]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(sample_means_n5, bins=30, color='salmon', edgecolor='black')
axes[0].set_title("Sample Means (n=5)")
axes[0].set_xlabel("Sample Mean")
axes[0].set_ylabel("Frequency")

axes[1].hist(sample_means_n30, bins=30, color='lightgreen', edgecolor='black')
axes[1].set_title("Sample Means (n=30)")
axes[1].set_xlabel("Sample Mean")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()

print("Std Dev of sample means (n=5):", np.std(sample_means_n5))
print("Std Dev of sample means (n=30):", np.std(sample_means_n30))
```

```python
# Task 5: CLT demonstration using random samples from a non-normal (exponential) distribution
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# Non-normal population: exponential distribution
population = np.random.exponential(scale=2, size=100000)

sample_means = [np.mean(np.random.exponential(scale=2, size=40)) for _ in range(1000)]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(population, bins=50, color='orange', edgecolor='black')
axes[0].set_title("Original Exponential Population")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Frequency")

axes[1].hist(sample_means, bins=30, color='cornflowerblue', edgecolor='black')
axes[1].set_title("Distribution of Sample Means (n=40)")
axes[1].set_xlabel("Sample Mean")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()

observation = ("Observation: The original population is heavily right-skewed (exponential), but the "
               "distribution of sample means is much more symmetric and bell-shaped, confirming the CLT.")
print(observation)
```
