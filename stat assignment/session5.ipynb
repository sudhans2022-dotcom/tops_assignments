```python
# Task 1: Delivery times (minutes) for 20 Zomato orders - calculate 25th, 50th, 75th percentiles
import numpy as np

delivery_times = [25, 30, 45, 20, 35, 40, 50, 28, 33, 38,
                   42, 22, 48, 31, 36, 29, 44, 27, 39, 46]

p25 = np.percentile(delivery_times, 25)
p50 = np.percentile(delivery_times, 50)
p75 = np.percentile(delivery_times, 75)

print("25th Percentile:", p25)
print("50th Percentile (Median):", p50)
print("75th Percentile:", p75)
```

```python
# Task 2: IPL player scores - divide into quartiles using numpy.quantile
import numpy as np

player_scores = [45, 78, 23, 90, 56, 34, 67, 12, 88, 50,
                  72, 39, 61, 95, 28]

Q1 = np.quantile(player_scores, 0.25)
Q2 = np.quantile(player_scores, 0.50)
Q3 = np.quantile(player_scores, 0.75)

print("Q1:", Q1)
print("Q2 (Median):", Q2)
print("Q3:", Q3)
```

```python
# Task 3: Daily step counts for one month - calculate Interquartile Range (IQR)
import numpy as np

daily_steps = [6500, 7200, 8000, 6000, 9000, 7500, 8200, 6800, 7100, 7600,
               8900, 6300, 7400, 8100, 6700, 7300, 8600, 6900, 7800, 7000,
               8300, 6600, 7900, 8400, 7200, 6400, 8800, 7500, 6900, 7700]

Q1 = np.percentile(daily_steps, 25)
Q3 = np.percentile(daily_steps, 75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Task 4: Detect outliers in Swiggy order amounts using the IQR method
import numpy as np

order_amounts = [200, 250, 300, 280, 320, 260, 240, 310, 270, 290,
                  900, 230, 255, 275, 50, 265, 285, 305, 245, 295]

def detect_outliers_iqr(data):
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = [x for x in data if x < lower_bound or x > upper_bound]
    return outliers

outliers = detect_outliers_iqr(order_amounts)
print("Outliers:", outliers)
```

```python
# Task 5: Boxplot of weekly screen time (hours) for the past 8 weeks - label axes and highlight outliers
import matplotlib.pyplot as plt

screen_time = [22, 25, 28, 24, 40, 26, 23, 27]

plt.figure(figsize=(6, 5))
plt.boxplot(screen_time, vert=True, patch_artist=True,
            flierprops=dict(marker='o', markerfacecolor='red', markersize=8, markeredgecolor='red'))
plt.title("Weekly Screen Time (Past 8 Weeks)")
plt.ylabel("Hours")
plt.xticks([1], ["Screen Time"])
plt.show()
```
