```python

#TASK 1-- Load a CSV file containing Flipkart product ratings and prices, then use Seaborn's jointplot with kind='scatter' to visualize the relationship between rating and price.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample Flipkart data
data = {
    "product": ["Phone", "Laptop", "Headphones", "Watch", "Tablet",
                "Camera", "TV", "Speaker", "Keyboard", "Mouse"],
    "rating": [4.5, 4.2, 4.0, 4.3, 4.1, 4.6, 4.4, 3.9, 4.2, 4.0],
    "price": [25000, 55000, 2500, 4000, 18000,
              45000, 60000, 3500, 1500, 800]
}

df = pd.DataFrame(data)

# Display first 5 rows
print(df.head())

# Jointplot
sns.jointplot(
    x="rating",
    y="price",
    data=df,
    kind="scatter"
)

plt.suptitle("Flipkart: Rating vs Price", y=1.02)
plt.show()
```

```python
#TASK 2-- Create a jointplot with kind='hex' to analyze the relationship between delivery time and order value using a dataset of Swiggy food orders.<br><br><em><strong>Hint:</strong> Use sns.jointplot(x='delivery_time', y='order_value', data=your_dataframe, kind='hex').</em>

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Swiggy order data
data = {
    "delivery_time": [
        20, 25, 30, 35, 40, 45, 25, 30, 35, 40,
        50, 55, 30, 35, 40, 45, 25, 30, 50, 60
    ],
    "order_value": [
        200, 250, 300, 350, 400, 450, 220, 280, 320, 390,
        500, 550, 270, 330, 420, 480, 230, 290, 520, 600
    ]
}

swiggy_df = pd.DataFrame(data)

# Hexbin jointplot
sns.jointplot(
    x="delivery_time",
    y="order_value",
    data=swiggy_df,
    kind="hex"
)

plt.suptitle("Swiggy: Delivery Time vs Order Value", y=1.02)
plt.show()
```

```python
#TASK 3-- Use Seaborn's jointplot with kind='kde' to visualize how the number of likes and comments are related in a dataset of Instagram posts.



# Sample Instagram post data
data = {
    "likes": [
        1000, 1500, 2000, 2500, 3000,
        3500, 4000, 4500, 5000, 6000,
        7000, 8000, 9000, 10000, 12000
    ],
    "comments": [
        50, 70, 100, 120, 150,
        180, 220, 250, 300, 350,
        400, 450, 500, 600, 700
    ]
}

instagram_df = pd.DataFrame(data)

# KDE jointplot
sns.jointplot(
    x="likes",
    y="comments",
    data=instagram_df,
    kind="kde"
)

plt.suptitle("Instagram: Likes vs Comments", y=1.02)
plt.show()
```
