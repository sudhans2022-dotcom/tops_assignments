```python

#TASK 1-- Use Seaborn's regplot to visualize the relationship between the number of likes and number of comments on a set of Instagram posts (create a small DataFrame with at least 10 sample data points).
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'likes': [120, 340, 89, 560, 210, 430, 670, 150, 720, 95, 310, 480],
    'comments': [10, 28, 7, 45, 18, 35, 52, 12, 60, 8, 25, 40]
}
df = pd.DataFrame(data)

sns.regplot(data=df, x='likes', y='comments')
plt.title('Likes vs Comments')
plt.show()
```

```python
#TASK 2-- Create a lmplot in Seaborn to show the regression line between food order price and delivery time using a sample Zomato-style dataset (minimum 15 rows with random values for price and delivery time).


np.random.seed(1)
df = pd.DataFrame({
    'price': np.random.randint(100, 1000, 15),
    'delivery_time': np.random.randint(15, 60, 15)
})

sns.lmplot(data=df, x='price', y='delivery_time')
plt.show()
```

```python
#TASK 4-- Generate a faceted regression plot (using lmplot with col parameter) to compare the relationship between hours spent on YouTube and exam scores, split by 'preparation method' (e.g., 'self-study' vs 'coaching') in your DataFrame.



np.random.seed(3)
df = pd.DataFrame({
    'youtube_hours': np.random.uniform(0, 8, 20),
    'exam_score': np.random.uniform(50, 100, 20),
    'prep_method': ['self-study']*10 + ['coaching']*10
})

sns.lmplot(data=df, x='youtube_hours', y='exam_score', col='prep_method')
plt.show()
```

```python
#TASK 5-- Modify a given lmplot that shows regression between Flipkart product price and number of reviews. Add a title, axis labels, and customize the marker color.<br><br><em><strong>Constraint:</strong> Do not change the regression logic—only update plot aesthetics.</em>

np.random.seed(4)
df = pd.DataFrame({
    'price': np.random.randint(5000, 80000, 15),
    'reviews': np.random.randint(10, 2000, 15)
})

sns.lmplot(data=df, x='price', y='reviews', scatter_kws={'color': 'green'}, line_kws={'color': 'black'})
plt.title('Flipkart Product Price vs Reviews')
plt.xlabel('Price')
plt.ylabel('Number of Reviews')
plt.show()
```
