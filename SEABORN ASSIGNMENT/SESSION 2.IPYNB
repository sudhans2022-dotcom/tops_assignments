```python

#TASK 1--Use sns.countplot() to visualize the number of orders placed by each payment method (e.g., UPI, Card, Cash) in a sample Swiggy orders DataFrame with at least 3 payment types.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Swiggy-style data
data = {
    'payment_method': ['UPI', 'Card', 'Cash', 'UPI', 'UPI', 'Card', 
                        'Cash', 'UPI', 'Card', 'Cash', 'UPI', 'Card']
}
df = pd.DataFrame(data)

plt.figure(figsize=(6, 4))
sns.countplot(x='payment_method', data=df, palette='Set2')
plt.title('Number of Orders by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.show()
```

```python
# TASK 2--Create a barplot using sns.barplot() to show the average delivery time (in minutes) for each food category (e.g., Pizza, Burger, Biryani, South Indian) from a Zomato-style dataset.<br><br><em><strong>Hint:</strong> Use the estimator parameter to show the mean delivery time for each category.</em>

data = {
    'food_category': ['Pizza', 'Burger', 'Biryani', 'South Indian', 'Pizza',
                       'Burger', 'Biryani', 'South Indian', 'Pizza', 'Biryani'],
    'delivery_time': [30, 20, 40, 25, 35, 22, 38, 28, 32, 42]
}
df = pd.DataFrame(data)

plt.figure(figsize=(6, 4))
sns.barplot(x='food_category', y='delivery_time', data=df, estimator='mean', palette='muted')
plt.title('Average Delivery Time by Food Category')
plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (mins)')
plt.show()
```

```python
#TASK 3--Modify your barplot to display the total number of orders instead of the average by setting estimator to sum in sns.barplot().

plt.figure(figsize=(6, 4))
sns.barplot(x='food_category', y='delivery_time', data=df, estimator='sum', palette='muted')
plt.title('Total Delivery Time by Food Category (Sum)')
plt.xlabel('Food Category')
plt.ylabel('Total Delivery Time (mins)')
plt.show()
```

```python
# TASK 4--Plot a countplot of movie genres from a BookMyShow-style dataset, but remove the confidence interval from the plot.<br><br><em><strong>Hint:</strong> Use the ci=None parameter to remove confidence intervals in sns.barplot().</em>

genre_counts = df['genre'].value_counts().reset_index()
genre_counts.columns = ['genre', 'count']

plt.figure(figsize=(6, 4))
sns.barplot(x='genre', y='count', data=genre_counts, estimator='sum', errorbar=None, palette='pastel')
plt.title('Number of Movies by Genre (No CI)')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.show()
```
