```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#TASK 1-- Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.

np.random.seed(42)


delivery_times = np.random.normal(loc=35, scale=8, size=40).round(1)
df = pd.DataFrame({'delivery_time': delivery_times})

plt.figure(figsize=(7, 4))
sns.histplot(df['delivery_time'], bins=10, color='orange', kde=False)
plt.title('Distribution of Swiggy Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Frequency')
plt.show()

```

```python
#TASK 2-- Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the distribution using sns.kdeplot(), showing the probability density curve.
np.random.seed(1)


days = pd.date_range(start='2026-08-01', periods=30, freq='D')
steps = np.random.normal(loc=7000, scale=1500, size=30).astype(int)
df_steps = pd.DataFrame({'date': days, 'steps': steps})

plt.figure(figsize=(7, 4))
sns.kdeplot(df_steps['steps'], fill=True, color='b')
plt.title('Probability Density of Daily Step Counts (1 Month)')
plt.xlabel('Step Count')
plt.ylabel('Density')
plt.show()

```

```python
#TASK 3-- Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:</strong> Use the same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>

np.random.seed(7)


spending = np.random.normal(loc=1800, scale=400, size=50).round(0)
df_spend = pd.DataFrame({'spending': spending})

plt.figure(figsize=(7, 4))
sns.histplot(df_spend['spending'], bins=12, color='skyblue', stat='density', label='Histogram')
sns.kdeplot(df_spend['spending'], color='red', linewidth=2, label='KDE')
plt.title('Distribution of Monthly Zomato Spending')
plt.xlabel('Spending (₹)')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python

#TASK 4-- Given a dataset of IPL match scores, use sns.histplot() to plot the run distribution. Change the number of bins and observe how the plot changes. Write 2-3 lines about what you notice.<br><br><em><strong>Hint:</strong> Try bins=5, bins=10, and bins=20.</em>
np.random.seed(3)


runs = np.random.normal(loc=165, scale=25, size=60).astype(int)
df_runs = pd.DataFrame({'runs': runs})

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(df_runs['runs'], bins=5, ax=axes[0], color='purple')
axes[0].set_title('bins=5')

sns.histplot(df_runs['runs'], bins=10, ax=axes[1], color='purple')
axes[1].set_title('bins=10')

sns.histplot(df_runs['runs'], bins=20, ax=axes[2], color='purple')
axes[2].set_title('bins=20')

plt.suptitle('IPL Run Distribution — Effect of Bin Count')
plt.tight_layout()
plt.show()
```
