```python

#TASK 1 Use sns.scatterplot() to visualize the relationship between the number of followers and the number of posts for 20 fictional Instagram users. Label the axes appropriately.

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 20 fictional Instagram users
data = {
    "user": [
        "User1", "User2", "User3", "User4", "User5",
        "User6", "User7", "User8", "User9", "User10",
        "User11", "User12", "User13", "User14", "User15",
        "User16", "User17", "User18", "User19", "User20"
    ],
    "followers": [
        1200, 2500, 5000, 8000, 12000,
        15000, 18000, 22000, 30000, 35000,
        40000, 45000, 50000, 60000, 70000,
        80000, 90000, 100000, 120000, 150000
    ],
    "posts": [
        50, 80, 120, 150, 180,
        200, 250, 300, 350, 400,
        450, 500, 550, 600, 650,
        700, 750, 800, 900, 1000
    ]
}

instagram_df = pd.DataFrame(data)

# Scatter plot
sns.scatterplot(
    x="followers",
    y="posts",
    data=instagram_df
)

plt.title("Instagram Followers vs Number of Posts")
plt.xlabel("Number of Followers")
plt.ylabel("Number of Posts")
plt.show()
```

```python
#TASK 2 Create a line plot using sns.lineplot() to show the daily number of food orders on Zomato for each day over a month. Use a list of 30 random order counts to simulate the data.



# Generate 30 random order counts
np.random.seed(42)

orders = np.random.randint(100, 500, 30)

# Create DataFrame
zomato_df = pd.DataFrame({
    "day": range(1, 31),
    "orders": orders
})

print(zomato_df)

# Line plot
sns.lineplot(
    x="day",
    y="orders",
    data=zomato_df,
    marker="o"
)

plt.title("Daily Zomato Food Orders - One Month")
plt.xlabel("Day of Month")
plt.ylabel("Number of Orders")
plt.xticks(range(1, 31))
plt.show()
```

```python
#TASK 3 Using sns.relplot(), plot the relationship between movie ratings and the number of reviews for different genres (Action, Comedy, Drama) based on a sample dataset of 15 movies. Use the 'hue' parameter to differentiate genres.<br><br><em><strong>Hint:</strong> Create a DataFrame with columns: 'movie', 'rating', 'reviews', and 'genre'.</em>


# Movie dataset
data = {
    "movie": [
        "Movie A", "Movie B", "Movie C", "Movie D", "Movie E",
        "Movie F", "Movie G", "Movie H", "Movie I", "Movie J",
        "Movie K", "Movie L", "Movie M", "Movie N", "Movie O"
    ],
    "rating": [
        8.2, 7.5, 8.8, 6.9, 7.8,
        8.5, 7.2, 9.0, 6.8, 8.0,
        7.7, 8.9, 7.0, 8.4, 7.6
    ],
    "reviews": [
        12000, 8000, 20000, 5000, 10000,
        15000, 6000, 25000, 4500, 11000,
        9000, 22000, 5500, 16000, 7500
    ],
    "genre": [
        "Action", "Comedy", "Drama", "Action", "Comedy",
        "Drama", "Action", "Drama", "Comedy", "Action",
        "Comedy", "Drama", "Action", "Drama", "Comedy"
    ]
}

movies_df = pd.DataFrame(data)

print(movies_df)

# Relplot
sns.relplot(
    x="rating",
    y="reviews",
    data=movies_df,
    hue="genre",
    kind="scatter"
)

plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
#TASK 4 Generate a regression plot using sns.regplot() to analyze the correlation between the number of hours spent on YouTube and exam scores for a group of 25 students. Interpret whether the trend is positive or negative.



# Generate sample data
np.random.seed(42)

hours = np.array([
    1, 2, 2.5, 3, 3.5,
    4, 4.5, 5, 5.5, 6,
    6.5, 7, 7.5, 8, 8.5,
    9, 9.5, 10, 10.5, 11,
    11.5, 12, 12.5, 13, 14
])

# Exam scores with a negative relationship
scores = np.array([
    92, 90, 88, 86, 85,
    83, 82, 80, 78, 76,
    75, 73, 71, 69, 67,
    65, 63, 61, 59, 57,
    55, 53, 51, 49, 47
])

students_df = pd.DataFrame({
    "youtube_hours": hours,
    "exam_score": scores
})

print(students_df)

# Regression plot
sns.regplot(
    x="youtube_hours",
    y="exam_score",
    data=students_df
)

plt.title("YouTube Hours vs Exam Scores")
plt.xlabel("Hours Spent on YouTube")
plt.ylabel("Exam Score")
plt.show()
```
