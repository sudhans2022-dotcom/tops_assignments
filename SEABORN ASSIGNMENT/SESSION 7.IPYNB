```python

# TASK 1-- Create a Pandas DataFrame with at least 5 columns representing different features of food delivery orders (e.g., delivery_time, order_value, tip_amount, rating, delivery_distance) and 20 rows of random data, then calculate and display the correlation matrix using the .corr() function.


np.random.seed(10)

n_orders = 20

df_delivery = pd.DataFrame({
    'delivery_time': np.random.randint(15, 60, n_orders),               # minutes
    'order_value': np.round(np.random.uniform(100, 1500, n_orders), 2), # currency
    'tip_amount': np.round(np.random.uniform(0, 100, n_orders), 2),     # currency
    'rating': np.random.randint(1, 6, n_orders),                        # 1-5 stars
    'delivery_distance': np.round(np.random.uniform(0.5, 15, n_orders), 2)  # km
})

print(df_delivery)

corr_matrix = df_delivery.corr()
print(corr_matrix.round(3))
```

```python

#TASK 4-- Take the Flipkart mobile phones dataset (or any small CSV with product features you can find online), compute the correlation matrix for at least 4 numeric columns (e.g., price, rating, RAM, battery_capacity), and generate a Seaborn heatmap with annotations and a diverging color palette.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Simple mobile phone dataset
data = {
    'price': [9999, 15999, 27999, 45999, 69999, 18999, 39999, 89999],
    'rating': [3.9, 4.1, 4.3, 4.5, 4.6, 4.0, 4.4, 4.8],
    'ram_gb': [3, 4, 6, 8, 12, 4, 8, 16],
    'battery_capacity': [4000, 5000, 4500, 4300, 4500, 5000, 4200, 5000]
}

df = pd.DataFrame(data)
print(df)

# Correlation matrix
corr_matrix = df.corr()
print(corr_matrix)

# Heatmap with diverging palette
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Mobile Phone Feature Correlation')
plt.show()
```
