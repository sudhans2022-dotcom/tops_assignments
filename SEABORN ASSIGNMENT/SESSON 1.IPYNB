```python
#task 1-- nstall Seaborn in your Python environment and create a new Jupyter notebook named seaborn_intro.ipynb. Import seaborn as sns and matplotlib.pyplot as plt, then display the versions of both libraries.

import seaborn as sns
import matplotlib.pyplot as plt

print("Seaborn Version:", sns.__version__)
print("Matplotlib Version:", plt.matplotlib.__version__)
```

```python
#task 2-- Using Seaborn's lineplot(), plot the daily temperature trend for one week using the following data: days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'], temps = [32, 34, 33, 35, 36, 37, 34]. Label the axes and title the chart 'Weekly Temperature Trend'.

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")
```

```python
# task 3--Recreate the above line plot using Seaborn's axes-level and figure-level functions separately. Briefly note one difference you observe in how these functions work.<br><br><em><strong>Hint:</strong> Use sns.lineplot() for axes-level and sns.relplot() for figure-level.</em>

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")
```

```python
# task 4--Apply each of Seaborn's built-in themes (darkgrid, whitegrid, ticks) to your temperature trend plot and save three separate images. Compare how the grid and background change with each theme.<br><br><em><strong>Hint:</strong> Use sns.set_theme(style='theme_name') before plotting.</em>
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.relplot(
    x=days,
    y=temps,
    kind='line',
    marker='o'
)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")
```

```python
#whitegrid
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.set_theme(style='darkgrid')

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Darkgrid")

plt.savefig("temperature_darkgrid.png")

plt.show()
```

```python
#ticks

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.set_theme(style='whitegrid')

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend - Whitegrid")

plt.savefig("temperature_whitegrid.png")

plt.show()
```
