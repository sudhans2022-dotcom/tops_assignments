```python

#task 1--Load the 'tips' dataset from seaborn and create a basic pairplot for all numerical columns.
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
 
# Load the built-in 'tips' dataset
tips = sns.load_dataset('tips')
 
# Basic pairplot for all numerical columns
sns.pairplot(tips)
plt.suptitle('Pairplot of Tips Dataset (All Numerical Columns)', y=1.02)
plt.savefig('task1_tips_pairplot.png', dpi=150, bbox_inches='tight')
print("Saved task1_tips_pairplot.png")
print(tips.head())
 
```

```python
#task 2--Use the pairplot function to visualize the 'tips' dataset, but this time set the hue parameter to 'sex' and observe how the data points are colored based on gender.

# Pairplot colored by gender
sns.pairplot(tips, hue='sex')
plt.suptitle('Pairplot of Tips Dataset (colored by Sex)', y=1.02)
plt.savefig('task2_tips_pairplot_sex.png', dpi=150, bbox_inches='tight')
print("Saved task2_tips_pairplot_sex.png/")
```

```python
#task 3--Create a pairplot for the 'tips' dataset with the hue set to 'smoker', and set the diag_kind parameter to 'kde' to show kernel density estimates on the diagonal plots.<br><br><em><strong>Hint:</strong> Use diag_kind='kde' in the pairplot function.</em>


# Pairplot colored by smoker status, with KDE curves on the diagonal
sns.pairplot(tips, hue='smoker', diag_kind='kde')
plt.suptitle('Pairplot of Tips Dataset (colored by Smoker, KDE diagonal)', y=1.02)
plt.savefig('task3_tips_pairplot_smoker_kde.png', dpi=150, bbox_inches='tight')
print("Saved task3_tips_pairplot_smoker_kde.png")
```

```python
#task 4--Pick any two numerical columns from the 'tips' dataset and use pairplot to visualize only those columns, setting hue to 'time' (Lunch/Dinner).<br><br><em><strong>Constraint:</strong> Use the vars parameter to select the columns.</em>
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Select only 'total_bill' and 'tip', colored by time (Lunch/Dinner)
sns.pairplot(tips, vars=['total_bill', 'tip'], hue='time')
plt.suptitle('Pairplot: Total Bill vs Tip (colored by Time)', y=1.02)
plt.savefig('task4_tips_pairplot_vars_time.png', dpi=150, bbox_inches='tight')
print("Saved task4_tips_pairplot_vars_time.png")
```
