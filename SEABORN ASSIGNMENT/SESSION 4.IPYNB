```python

#TASK 1-- Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('flipkart_prices.csv')

# Visualize price distribution
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x='product_category', y='price')
plt.title('Flipkart Product Price Distribution by Category')
plt.xlabel('Product Category')
plt.ylabel('Price (₹)')
plt.yscale('log')  # helpful if prices span a wide range (e.g. Electronics vs Books)
plt.tight_layout()
plt.show()
```

```python
#TASK 2--Use sns.violinplot() to compare the order amounts for two different Zomato restaurant categories (e.g., 'Fast Food' vs 'Desserts') from a sample dataset.


# Sample dataset — replace with your real Zomato data
data = {
    'restaurant_category': ['Fast Food']*5 + ['Desserts']*5,
    'order_amount': [220, 340, 410, 180, 500, 150, 220, 280, 300, 260]
}
df = pd.DataFrame(data)

# Filter to just the two categories you want to compare
subset = df[df['restaurant_category'].isin(['Fast Food', 'Desserts'])]

plt.figure(figsize=(7, 5))
sns.violinplot(data=subset, x='restaurant_category', y='order_amount', 
               hue='restaurant_category', palette='Set2', legend=False)
plt.title('Zomato Order Amount: Fast Food vs Desserts')
plt.xlabel('Restaurant Category')
plt.ylabel('Order Amount (₹)')
plt.tight_layout()
plt.show()
```
