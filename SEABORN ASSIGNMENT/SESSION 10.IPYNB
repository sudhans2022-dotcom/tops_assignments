```python

#TASK 1-- Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of songs streamed per day on Spotify, and observe how the plot elements change compared to the default context.

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'songs_streamed': [45, 60, 38, 72, 90, 110, 95]
}
df = pd.DataFrame(data)

sns.set()
sns.set_context('poster')

plt.figure(figsize=(10, 7))
sns.barplot(data=df, x='day', y='songs_streamed')
plt.title('Songs Streamed per Day on Spotify')
plt.xlabel('Day')
plt.ylabel('Songs Streamed')
plt.show()


```

```python
#TASK 2-- Create a Seaborn countplot showing the distribution of food orders by cuisine type (e.g., Indian, Chinese, Italian) using a 'colorblind' palette, similar to how Zomato might visualize order trends.<br><br><em><strong>Hint:</strong> Use palette='colorblind' in your plot function.</em>


data = {
    'cuisine': ['Indian', 'Chinese', 'Italian', 'Indian', 'Mexican',
                'Chinese', 'Indian', 'Italian', 'Indian', 'Mexican',
                'Chinese', 'Indian', 'Italian', 'Chinese', 'Indian',
                'Mexican', 'Italian', 'Indian', 'Chinese', 'Indian']
}
df = pd.DataFrame(data)

sns.countplot(data=df, x='cuisine', hue='cuisine', palette='colorblind', legend=False)
plt.title('Food Orders by Cuisine Type')
plt.xlabel('Cuisine')
plt.ylabel('Number of Orders')
plt.show()
```

```python
#TASK 3-- Generate a Seaborn lineplot of daily step counts (simulate data for 2 weeks) and style it using the 'muted' palette and 'notebook' context for a clean, presentation-ready look.

np.random.seed(15)

days = list(range(1, 15))  # 2 weeks
steps = np.random.randint(3000, 12000, 14)

df = pd.DataFrame({'day': days, 'steps': steps})

sns.set_theme(context='notebook', palette='muted')

plt.figure(figsize=(9, 6))
sns.lineplot(data=df, x='day', y='steps', marker='o')
plt.title('Daily Step Count Over 2 Weeks')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.show()
```

```python
#TASK 4-- Refactor an existing Seaborn plot by switching its palette from 'deep' to 'colorblind' and changing the context from 'talk' to 'poster'. Briefly describe the visual differences you notice after applying these changes.
data = {'category': ['A', 'B', 'C', 'D', 'E'], 'value': [23, 45, 12, 38, 29]}
df = pd.DataFrame(data)


sns.set_theme(context='talk', palette='deep')
sns.barplot(data=df, x='category', y='value', hue='category', legend=False)
plt.title('Before: talk + deep')
plt.show()

# After
sns.set_theme(context='poster', palette='colorblind')
sns.barplot(data=df, x='category', y='value', hue='category', legend=False)
plt.title('After: poster + colorblind')
plt.show()
```

```python
#TASK 5-- Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use sns.set() to set a 'deep' palette, 'talk' context, and customize the font scale and grid style. Apply your theme to a barplot showing sales of 5 product categories

data = {
    'category': ['Electronics', 'Fashion', 'Home & Kitchen', 'Beauty', 'Grocery'],
    'sales': [82000, 65000, 47000, 39000, 58000]
}
df = pd.DataFrame(data)

sns.set_theme(context='talk', palette='deep', style='whitegrid', font_scale=1.1)

plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='category', y='sales', hue='category', legend=False)
plt.title('Flipkart-Style Sales Dashboard: Product Category Sales')
plt.xlabel('Product Category')
plt.ylabel('Sales (₹)')
plt.show()

```
