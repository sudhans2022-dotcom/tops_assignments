```python

#TASK 1-- Create a Seaborn barplot showing the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and save the chart as a PNG file with 300 DPI.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample food order data
data = {
    "category": ["Pizza", "Burger", "Biryani", "Momos", "Ice Cream"],
    "orders": [450, 380, 520, 300, 250]
}

df = pd.DataFrame(data)

# Create barplot
plt.figure(figsize=(8, 5))

sns.barplot(
    x="category",
    y="orders",
    data=df
)

plt.title("Number of Orders by Food Category")
plt.xlabel("Food Category")
plt.ylabel("Number of Orders")

# Save as PNG with 300 DPI
plt.savefig(
    "food_orders.png",
    dpi=300,
    format="png",
    bbox_inches="tight"
)

plt.show()
```

```python
#TASK 2-- Generate a Seaborn line plot visualizing daily step counts for one week, then export the chart as an SVG file suitable for use in a PowerPoint presentation.<br><br><em><strong>Hint:</strong> Use plt.savefig('filename.svg', format='svg').</em>


# One week step-count data
data = {
    "day": [
        "Monday", "Tuesday", "Wednesday",
        "Thursday", "Friday", "Saturday", "Sunday"
    ],
    "steps": [
        6500, 7200, 8000, 7500, 9000, 11000, 9500
    ]
}

steps_df = pd.DataFrame(data)

# Create line plot
plt.figure(figsize=(9, 5))

sns.lineplot(
    x="day",
    y="steps",
    data=steps_df,
    marker="o"
)

plt.title("Daily Step Counts - One Week")
plt.xlabel("Day")
plt.ylabel("Number of Steps")

# Export as SVG
plt.savefig(
    "weekly_steps.svg",
    format="svg",
    bbox_inches="tight"
)

plt.show()
```

```python
#TASK 3-- Customize a Seaborn chart using the 'darkgrid' style and a color palette of your choice, then save it as a high-resolution PNG image with 1200 DPI.


# Set Seaborn style
sns.set_style("darkgrid")

# Choose color palette
sns.set_palette("Set2")

# Sample sales data
data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [25000, 28000, 32000, 30000, 38000, 42000]
}

sales_df = pd.DataFrame(data)

# Create chart
plt.figure(figsize=(9, 5))

sns.barplot(
    x="month",
    y="sales",
    data=sales_df
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

# Save at 1200 DPI
plt.savefig(
    "monthly_sales_1200dpi.png",
    dpi=1200,
    format="png",
    bbox_inches="tight"
)

plt.show()
```

```python
#TASK 4-- Create a template function called apply_presentation_theme() that applies a consistent Seaborn style, font, and color palette to any chart, then use it to style and export a pie chart showing your top 5 most-used apps.



# Function for presentation theme
def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        font="DejaVu Sans",
        palette="Set2"
    )


# Apply theme
apply_presentation_theme()


# Top 5 most-used apps
apps = [
    "Instagram",
    "YouTube",
    "WhatsApp",
    "Spotify",
    "Google"
]

usage = [
    35,
    25,
    20,
    12,
    8
]


# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    usage,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Most-Used Apps")

# Save high-resolution PNG
plt.savefig(
    "top_5_apps.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
