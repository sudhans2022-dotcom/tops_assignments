```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

#TASK 1-- Create a line chart showing the number of daily active users on a music streaming app (like Spotify) over 7 days, and set the figure size to 10 inches wide by 4 inches tall using plt.figure().


days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
users = [500, 600, 550, 700, 750, 900, 850]

plt.figure(figsize=(10, 4))

plt.plot(days, users)

plt.xlabel("Days")
plt.ylabel("Active Users")
plt.title("Daily Active Users")

plt.show()
```

```python
#TASK 2-- Plot two lines on the same graph representing the number of food orders on Zomato and Swiggy over a week, using different line styles (e.g., dashed for Zomato, dotted for Swiggy) and add a legend to distinguish them.


days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [20, 25, 22, 30, 35, 40, 38]
swiggy = [15, 20, 18, 25, 30, 35, 32]

plt.plot(days, zomato, linestyle="--", label="Zomato")
plt.plot(days, swiggy, linestyle=":", label="Swiggy")

plt.xlabel("Days")
plt.ylabel("Food Orders")
plt.title("Zomato vs Swiggy Orders")

plt.legend()

plt.show()
```

```python
#TASK 3-- Apply the 'ggplot' style to a chart showing Flipkart's daily sales for 10 days using plt.style.use('ggplot'), and customize the line color to orange.

plt.style.use("ggplot")

days = ["Day1", "Day2", "Day3", "Day4", "Day5",
        "Day6", "Day7", "Day8", "Day9", "Day10"]

sales = [100, 150, 130, 180, 200, 220, 190, 250, 280, 300]

plt.plot(days, sales, color="orange")

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Flipkart Daily Sales")

plt.show()
```

```python
#TASK 4-- Choose a color palette that matches Instagram's branding and plot a bar chart showing the number of posts uploaded each day for a week; use those colors for the bars.<br><br><em><strong>Hint:</strong> You can use a list of hex color codes to set bar colors.</em>

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
posts = [5, 8, 6, 10, 12, 15, 9]

colors = [
    "#833AB4",
    "#C13584",
    "#E1306C",
    "#FD1D1D",
    "#F56040",
    "#FCAF45",
    "#405DE6"
]

plt.bar(days, posts, color=colors)

plt.xlabel("Days")
plt.ylabel("Number of Posts")
plt.title("Instagram Posts Per Day")

plt.show()
```

```python
#TASK 5-- Take any one of your previous charts and restyle it to match a dark theme using plt.style.use('dark_background'), then adjust the line or bar colors so that all text and data remain clearly visible.<br><br><em><strong>Constraint:</strong> Do not use the same colors as in the original chart.</em>


plt.style.use("dark_background")

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

users = [500, 600, 550, 700, 750, 900, 850]

plt.plot(days, users, color="cyan")

plt.xlabel("Days")
plt.ylabel("Active Users")
plt.title("Daily Active Users - Dark Theme")

plt.show()
```
