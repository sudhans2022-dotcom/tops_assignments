```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

#TASK 1-- Use plt.subplot() to create a figure with 2 rows and 1 column of plots showing your daily step count and water intake (in litres) for the last 7 days. Plot both as line charts, one in each subplot.


days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [5000, 6000, 7500, 8000, 6500, 9000, 7000]
water = [2, 2.5, 2, 3, 2.5, 3, 2.5]

plt.subplot(2, 1, 1)

plt.plot(days, steps)
plt.title("Daily Steps")
plt.ylabel("Steps")


plt.subplot(2, 1, 2)

plt.plot(days, water)
plt.title("Water Intake")
plt.xlabel("Days")
plt.ylabel("Litres")

plt.tight_layout()
plt.show()

```

```python
#TASK 2-- Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of Instagram posts you made in each quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.


quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [20, 30, 25, 40]

plt.subplots(2, 2)


plt.subplot(2, 2, 1)
plt.plot(quarters, posts)
plt.title("Line Chart")



plt.subplot(2, 2, 2)
plt.bar(quarters, posts)
plt.title("Bar Chart")



plt.subplot(2, 2, 3)
plt.scatter(quarters, posts)
plt.title("Scatter Chart")



plt.subplot(2, 2, 4)
plt.pie(posts, labels=quarters, autopct="%1.1f%%")
plt.title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
#TASK 3-- Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and Domino's over the last 6 months, one app per subplot. Share the y-axis among all subplots so you can easily compare order counts across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>


months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

zomato = [5, 7, 6, 8, 10, 9]
swiggy = [4, 6, 5, 7, 8, 10]
dominos = [2, 3, 4, 3, 5, 4]

fig, ax = plt.subplots(1, 3, sharey=True)

ax[0].plot(months, zomato)
ax[0].set_title("Zomato")
ax[0].set_ylabel("Orders")

ax[1].plot(months, swiggy)
ax[1].set_title("Swiggy")

ax[2].plot(months, dominos)
ax[2].set_title("Domino's")

plt.tight_layout()
plt.show()
```

```python
#TASK 4-- Arrange a 2x2 grid of subplots using plt.subplots() to visualize your monthly mobile data usage (in GB) for four different apps: YouTube, Instagram, Spotify, and WhatsApp. Ensure all subplots share the same x-axis (months) for easy comparison.<br><br><em><strong>Hint:</strong> Use the 'sharex' parameter in plt.subplots().</em>


months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

youtube = [10, 12, 15, 14, 18, 20]
instagram = [5, 7, 6, 8, 9, 10]
spotify = [3, 4, 5, 4, 6, 7]
whatsapp = [2, 3, 2, 4, 3, 5]

fig, ax = plt.subplots(2, 2, sharex=True)


ax[0, 0].plot(months, youtube)
ax[0, 0].set_title("YouTube")
ax[0, 0].set_ylabel("GB")

ax[0, 1].plot(months, instagram)
ax[0, 1].set_title("Instagram")

ax[1, 0].plot(months, spotify)
ax[1, 0].set_title("Spotify")
ax[1, 0].set_ylabel("GB")


ax[1, 1].plot(months, whatsapp)
ax[1, 1].set_title("WhatsApp")

plt.tight_layout()
plt.show()
```
