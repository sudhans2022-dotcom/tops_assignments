```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#TASK1 Use plt.bar() to create a vertical bar chart showing the number of daily orders for 5 popular food delivery apps (e.g., Zomato, Swiggy, Uber Eats, Dunzo, Domino's) and set a custom color for each bar.
delivery_apps=["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
daily_orders=[120,152,185,172,128]
custom_colors = ['#E23744', '#FC8019', '#5FB709', '#006491', '#E31837']

plt.bar(delivery_apps,daily_orders,color=custom_colors)
plt.grid(axis="y",color="gray",linestyle=":",linewidth=0.8)
plt.title("daily orders",fontsize=20)
plt.xlabel("delivery apps", fontsize=20)
plt.ylabel("daily orders", fontsize=20)
plt.show()
```

```python
#TASK 2--Create a horizontal bar chart using plt.barh() to display the number of followers for 6 different Instagram influencers; add a title and label the x-axis as 'Followers (in millions)'.
influencers = ["Virat Kohli", "Cristiano Ronaldo", "Selena Gomez",
               "Dwayne Johnson", "Ariana Grande", "Kim Kardashian"]

followers = [270, 650, 420, 395, 380, 360]

plt.figure(figsize=(10, 6))
plt.barh(influencers,followers)
plt.title("Instagram Influencers and Their Followers",fontsize=20)
plt.xlabel("Followers (in millions)",fontsize=20)
plt.ylabel("Influencers",fontsize=20)

plt.show()
```

```python
# TASK 3- Modify your vertical bar chart to display the exact order count as a text label above each bar using plt.text().<br><br><em><strong>Hint:</strong> Use a for loop to iterate through the bars and place the text labels at the correct height.</em>

months = ["Jan", "Feb", "Mar", "Apr", "May"]
orders = [120, 180, 150, 220, 200]

plt.bar(months, orders)

plt.title("Monthly Order Count")
plt.xlabel("Month")
plt.ylabel("Order Count")


for i, value in enumerate(orders):
    plt.text(i, value, str(value), ha="center", va="bottom")

plt.show()
```

```python
# TASK 4-Change the width of the bars in your food delivery app chart to make them thinner, and adjust the spacing so the bars do not overlap.<br><br><em><strong>Constraint:</strong> Use the 'width' parameter in plt.bar() and experiment with values less than 0.8.</em>


apps = ["Zomato", "Swiggy", "Uber Eats", "Foodpanda"]
orders = [500, 450, 300, 200]

plt.figure(figsize=(8, 5))

plt.bar(apps, orders, width=0.5)

plt.grid()

plt.title("Food Delivery App Orders")
plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")

plt.show()
```
