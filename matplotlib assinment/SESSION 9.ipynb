```python
import numpy as np 
import pandas as pll
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

#TASK 1-- Create a stacked bar chart using Matplotlib that shows the number of food orders from Zomato and Swiggy for each day of the week. Use different colors for each platform and add a legend.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [20, 25, 30, 22, 35, 40, 38]
swiggy = [15, 20, 25, 18, 30, 35, 32]

plt.bar(days, zomato, label="Zomato")
plt.bar(days, swiggy, bottom=zomato, label="Swiggy")

plt.xlabel("Days")
plt.ylabel("Food Orders")
plt.title("Zomato and Swiggy Food Orders")

plt.legend()
plt.show()
```

```python
#TASK 2-- Plot an area chart that visualizes the daily active users on Spotify and YouTube over a 7-day period using Matplotlib's fill_between(). Make sure each platform's area is filled with a different color and add labels for clarity.



days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [500, 600, 550, 700, 750, 900, 850]
youtube = [700, 750, 800, 850, 900, 1000, 950]

plt.plot(days, spotify, label="Spotify")
plt.fill_between(days, spotify, alpha=0.5)

plt.plot(days, youtube, label="YouTube")
plt.fill_between(days, youtube, alpha=0.5)

plt.xlabel("Days")
plt.ylabel("Active Users")
plt.title("Spotify vs YouTube Active Users")

plt.legend()
plt.show()
```

```python
#TASK 3-- Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard deviation for each day's delivery time using Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr parameter to show error bars for each data point.</em>


days = ["Day1", "Day2", "Day3", "Day4", "Day5",
        "Day6", "Day7", "Day8", "Day9", "Day10"]

delivery_time = [35, 40, 32, 45, 38, 42, 30, 36, 40, 34]

std = [3, 4, 2, 5, 3, 4, 2, 3, 4, 2]

plt.errorbar(days, delivery_time, yerr=std, fmt="o")

plt.xlabel("Days")
plt.ylabel("Delivery Time (Minutes)")
plt.title("Flipkart Delivery Time")

plt.show()
```

```python
#TASK 4 -- Plot a time series line chart showing your daily Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to display dates in 'DD MMM' format (e.g., '12 Jun').

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta

dates = []

start = datetime(2026, 8, 23)

for i in range(14):
    dates.append(start + timedelta(days=i))

balance = [5000, 4800, 5200, 4500, 4700, 5000, 5500,
           5300, 5800, 5600, 6000, 5700, 6200, 6500]

plt.plot(dates, balance)

plt.xlabel("Date")
plt.ylabel("Wallet Balance (₹)")
plt.title("Paytm Wallet Balance")

plt.gca().xaxis.set_major_formatter(
    mdates.DateFormatter("%d %b")
)

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
