```python

#TASK 1-- Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.

import matplotlib.pyplot as plt

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

users = [12000, 13500, 12800, 14500, 15200, 16800,
         17500, 18200, 19500, 21000, 22500, 21800]

# Find highest user count
max_users = max(users)
max_index = users.index(max_users)

plt.figure(figsize=(10, 5))

plt.plot(months, users, marker='o', linewidth=2)

plt.annotate(
    f"Highest Users: {max_users}",
    xy=(months[max_index], max_users),
    xytext=(months[max_index - 2], max_users + 2000),
    arrowprops=dict(arrowstyle="->"),
    fontsize=10
)

plt.title("Monthly Active Users - Music Streaming App")
plt.xlabel("Month")
plt.ylabel("Daily Active Users")
plt.grid(True)

plt.show()
```

```python
#TASK 2-- Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>



months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

orders = [4500, 5200, 4800, 6100, 5800, 6500,
          7200, 6800, 5500, 4900, 4300, 5100]


min_orders = min(orders)
min_index = orders.index(min_orders)

plt.figure(figsize=(10, 5))

plt.bar(months, orders)

plt.annotate(
    f"Lowest Orders: {min_orders}",
    xy=(months[min_index], min_orders),
    xytext=(months[min_index], min_orders + 1500),
    arrowprops=dict(
        arrowstyle="->",
        connectionstyle="arc3"
    ),
    ha="center"
)

plt.title("Monthly Orders - Food Delivery App")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.show()
```

```python
#TASK 3-- Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.


payment_methods = ["UPI", "Card", "Wallet", "Cash"]
payments = [50, 25, 15, 10]

plt.figure(figsize=(7, 7))

plt.pie(
    payments,
    labels=payment_methods,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Payment Methods Used - E-commerce App")

plt.legend(
    payment_methods,
    loc="lower right",
    title="Payment Methods"
)

plt.show()
```

```python
#TASK 4-- Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>



movies = [
    "Movie A", "Movie B", "Movie C", "Movie D",
    "Movie E", "Movie F", "Movie G", "Movie H"
]

ratings = [7.2, 8.1, 6.8, 9.2, 7.5, 8.7, 6.5, 8.0]

reviews = [1200, 2500, 900, 4500, 1800, 3200, 700, 2100]

# Find highest rating
max_rating = max(ratings)
max_index = ratings.index(max_rating)

plt.figure(figsize=(9, 6))

plt.scatter(reviews, ratings, s=100)

plt.annotate(
    f"{movies[max_index]}\nHighest Rating: {max_rating}",
    xy=(reviews[max_index], ratings[max_index]),
    xytext=(reviews[max_index] + 500, ratings[max_index] - 0.8),
    arrowprops=dict(arrowstyle="->"),
    bbox=dict(
        boxstyle="round,pad=0.5",
        edgecolor="black",
        facecolor="white"
    )
)

plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Number of Reviews")
plt.ylabel("Movie Rating")
plt.grid(True)

plt.show()
```
