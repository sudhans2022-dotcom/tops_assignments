```python

#TASK 1-- Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.
import matplotlib.pyplot as plt

restaurants = ["Domino's", "Burger King", "Biryani Blues", "Haldiram's", "Pizza Hut"]
orders = [120, 95, 80, 110, 75]

plt.figure(figsize=(8, 5))
plt.bar(restaurants, orders)

plt.title("Number of Orders - Zomato Restaurants")
plt.xlabel("Restaurants")
plt.ylabel("Number of Orders")
plt.xticks(rotation=20)

plt.tight_layout()


plt.savefig("zomato_orders.png", dpi=150)

plt.show()
```

```python
#TASK 2-- Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig().<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG.</em>
import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

steps = [6500, 7200, 8000, 7500, 9000, 10500, 8500]

plt.figure(figsize=(8, 5))
plt.plot(days, steps, marker="o")

plt.title("Daily Step Count")
plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.xticks(rotation=20)

plt.tight_layout()

# Save as SVG
plt.savefig("daily_steps.svg")

plt.show()
```

```python
#TASK 3-- Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().
import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
time_spent = [2.5, 3, 1.5, 1]

plt.figure(figsize=(7, 7))

plt.pie(
    time_spent,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Daily App Usage")

# Save as PDF
plt.savefig("app_usage.pdf")

plt.show()
```

```python
#TASK 5-- Export a Matplotlib chart as a PDF, then verify the PDF opens correctly and the chart looks sharp (not blurry) at 200% zoom. If the image looks pixelated, try exporting again with a higher DPI setting and compare the results.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 210, 195]

plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.tight_layout()

# Export PDF with 200 DPI
plt.savefig("monthly_sales_200dpi.pdf", dpi=200)

plt.show()
```
