```python
#TASK 2-- Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>



friends = [
    "Rahul", "Amit", "Priya", "Neha",
    "Karan", "Riya", "Ankit", "Pooja",
    "Vikas", "Sneha", "Rohit", "Simran"
]

amount_spent = [1200, 2500, 1800, 3200, 1500, 4000,
                2200, 2800, 3500, 1700, 3000, 2400]


orders = [5, 8, 6, 12, 5, 15,
          9, 10, 14, 6, 11, 8]


plt.scatter(orders, amount_spent)


for i in range(len(friends)):
    plt.annotate(
        friends[i],
        (orders[i], amount_spent[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Number of Zomato Orders")
plt.ylabel("Amount Spent (₹)")
plt.title("Zomato Orders vs Amount Spent")
plt.grid(True)

plt.show()
```

```python
#TASK 3-- Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>



orders = [5, 8, 6, 12, 5, 15, 9, 10, 14, 6, 11, 8]

amount = [1200, 2500, 1800, 3200, 1500, 4000,
          2200, 2800, 3500, 1700, 3000, 2400]

plt.scatter(orders, amount)

m, c = np.polyfit(orders, amount, 1)

x = np.array(orders)
y = m * x + c

plt.plot(x, y)

plt.xlabel("Number of Orders")
plt.ylabel("Amount Spent")
plt.title("Zomato Orders with Trend Line")

plt.show()

print("Slope =", m)
print("Intercept =", c)
```
