```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#TASK 1-Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. Set the line color to green and linewidth to 2.
daily_steps=[8550,7890,8100,7000,6900,5600,6948]
plt.plot(daily_steps,
        color="green",
        linewidth=2,
        marker="x",
        markerfacecolor="red"
        
        )
plt.grid()
plt.show()
```

```python
#TASK 2--Plot a line chart of your favorite music app's monthly active users for 6 months using plt.plot(). Add circle markers at each data point and set the marker color to red.<br><br><em><strong>Hint:</strong> Use the marker and markerfacecolor parameters in plt.plot().</em>
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
maus = [60245, 20878, 81545, 6820, 62896, 68632]
plt.figure(figsize=(8, 5))


plt.plot(months, maus, color='skyblue', linewidth=2, marker='o', markerfacecolor='red', markersize=8,linestyle="--")


plt.title('Spotify Monthly Active Users (MAUs) - 6 Month Trend', fontsize=14, fontweight='bold')
plt.xlabel('Months', fontsize=12)
plt.ylabel('Users (in Millions)', fontsize=12)
plt.grid(linestyle='--', alpha=0.7)


plt.show()
```

```python
# TASK 3--Create a line chart comparing the number of orders you placed on Zomato and Swiggy over the last 5 months. Use different line styles and colors for each app, and add a legend to distinguish them.
months = ['Apr', 'May', 'Jun', 'Jul', 'Aug']
zomato_orders = [12, 15, 8, 14, 19]
swiggy_orders = [10, 11, 14, 9, 16]

plt.plot(months,zomato_orders,
         color="black",
         marker="o",
         linestyle="--",
         linewidth=2,
         label="zomato_orders",
         markerfacecolor="b")

plt.plot(months,swiggy_orders,
         color="red",
         marker="x",
         linestyle="-",
         linewidth=2,
         label="swiggy_orders",
         markerfacecolor="yellow")

plt.legend()
plt.grid(axis='both', color='gray', linestyle='--', linewidth=1.5, alpha=0.7)
plt.title("placedd orders", fontsize=20)
plt.xlabel("month",fontsize=15)
plt.ylabel("no. of orders",fontsize=15)


plt.show
```

```python
# TASK 4--Add a main title 'Monthly Data Usage Trend' and label the X-axis as 'Month' and Y-axis as 'Data Used (GB)' to a line chart showing your mobile data usage for the past 6 months.<br><br><em><strong>Hint:</strong> Use plt.title(), plt.xlabel(), and plt.ylabel().</em>
months = ['Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug']
data_used = [45, 52, 48, 65, 60, 72]
plt.plot(months,data_used,
        color="b",
        linestyle="--",
        marker="o",
        markerfacecolor="y")
plt.grid(color="g",linestyle="--")
plt.xlabel("months",)
plt.ylabel("data_used (GB)")
plt.title("Monthly Data Usage Trend ",fontsize=20)
```
