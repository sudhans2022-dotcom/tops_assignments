```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#TASK 1-- Use plt.pie() to visualize the percentage of time you spend on 5 different apps in a day (e.g., Instagram, YouTube, WhatsApp, Zomato, Spotify) and display the percentage values on each segment using the autopct parameter.


apps = ['Instagram', 'YouTube', 'WhatsApp', 'Zomato', 'Spotify']
time_spent = [2, 3, 1.5, 1, 1.5]

plt.figure(figsize=(7, 7))

plt.pie(
    time_spent,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Daily App Usage')

plt.show()
```

```python

#TASK 2-- Create a pie chart showing the distribution of your monthly online spending across 4 categories: Food Delivery, Shopping, Entertainment, and UPI Payments, and use the explode parameter to highlight the category with the highest spending.<br><br><em><strong>Hint:</strong> Set a higher explode value for the segment you want to pull out from the pie.</em>

categories = [
    'Food Delivery',
    'Shopping',
    'Entertainment',
    'UPI Payments'
]

spending = [3000, 5000, 2000, 4000]

explode = [0, 0.1, 0, 0]

plt.figure(figsize=(7, 7))

plt.pie(
    spending,
    labels=categories,
    autopct='%1.1f%%',
    explode=explode,
    startangle=90
)

plt.title('Monthly Online Spending')

plt.show()
```

```python
#TASK 3-- Transform your pie chart into a donut chart to show the same spending data by adding a white circle at the center using plt.Circle and ax.add_artist.


categories = [
    'Food Delivery',
    'Shopping',
    'Entertainment',
    'UPI Payments'
]

spending = [3000, 5000, 2000, 4000]

explode = [0, 0.1, 0, 0]

fig, ax = plt.subplots(figsize=(7, 7))

ax.pie(
    spending,
    labels=categories,
    autopct='%1.1f%%',
    explode=explode,
    startangle=90
)

# Create white circle in the center
centre_circle = plt.Circle(
    (0, 0),
    0.70,
    fc='white'
)

ax.add_artist(centre_circle)

ax.set_title('Monthly Online Spending - Donut Chart')

plt.show()
```

```python
#TASK 4-- Suppose you want to compare the market share of 4 IPL cricket teams based on the number of Instagram followers. Build a donut chart where each segment shows the team name and follower percentage, and the chart title clearly states 'IPL Teams Instagram Followers Share'.<br><br><em><strong>Constraint:</strong> Use both labels and autopct to display information on the chart.</em>


teams = [
    'Mumbai Indians',
    'Chennai Super Kings',
    'Royal Challengers Bengaluru',
    'Kolkata Knight Riders'
]

followers = [
    14,
    16,
    18,
    8
]

fig, ax = plt.subplots(figsize=(8, 8))

ax.pie(
    followers,
    labels=teams,
    autopct='%1.1f%%',
    startangle=90
)

# Create the donut hole
centre_circle = plt.Circle(
    (0, 0),
    0.70,
    fc='white'
)

ax.add_artist(centre_circle)

ax.set_title(
    'IPL Teams Instagram Followers Share',
    fontsize=14
)

plt.show()
```
